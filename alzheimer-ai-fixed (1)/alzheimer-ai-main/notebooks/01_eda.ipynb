```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(PROJECT_ROOT + 'data/oasis_longitudinal.csv')

print("Shape:", df.shape)
print(df.head())

print(df['Group'].value_counts())
print(df['CDR'].value_counts())
```

```python
missing = df.isnull().sum()
print("Missing values:\n", missing[missing > 0])

for col in ['MMSE', 'nWBV', 'Age', 'EDUC']:
    n_miss = df[col].isnull().sum()

    if n_miss > 0:
        group_med = df.groupby('Group')[col].transform('median')
        df[col] = df[col].fillna(group_med).fillna(df[col].median())
        print(f"{col}: {n_miss} missing → group-median imputation applied")
```

```python
def map_label(row):
    if row['Group'] == 'Nondemented':
        return 'CN'
    elif row['CDR'] == 0.5:
        return 'MCI'
    else:
        return 'MA'

df['Label'] = df.apply(map_label, axis=1)

label_map = {
    'CN': 0,
    'MCI': 1,
    'MA': 2
}

df['y'] = df['Label'].map(label_map)

print(df['Label'].value_counts())
print(df[['Group', 'CDR', 'Label', 'y']].head(20))
```

```python
output_path = PROJECT_ROOT + 'data/processed/oasis_labeled.csv'
df.to_csv(output_path, index=False)

print("Saved:", output_path)
```
